```python
#TH 2.3
luong_gio = 100000
while True:
    try:
        gio_lam = float(input("So gio lam: "))
        if gio_lam < 0:
            print("Loi, So gio lam khong duoc am.")
            continue
        break
    except ValueError:
        print("Loi, Ban phai nhap du lieu so.")
if gio_lam <= 40:
    luong = gio_lam * luong_gio
else:
    luong = 40 * luong_gio + (gio_lam - 40) * luong_gio * 1.5
print(f"Tien luong: {luong:,.0f} dong")
```

```python
#TH 2.4
tong = 0
dem = 0
lon = float("-inf")
nho = float("inf")
while True:
    s = input("Nhap so (hoac 'xong'): ")
    if s == "xong":
        break
    try:
        x = float(s)
        tong += x
        dem += 1
        if x > lon:
            lon = x
        if x < nho:
            nho = x
    except ValueError:
        print("Gia tri khong hop le:", s)
if dem == 0:
    print("Chua co du lieu")
else:
    trung_binh = tong / dem
    print("So gia tri hop le:", dem)
    print("Tong:", tong)
    print("Trung binh:", trung_binh)
    print("Gia tri lon nhat:", lon)
    print("Gia tri nho nhat:", nho)
```

```python
#TH 2.7
n = 27
dem = 0
lon_nhat = n
while n!= 1:
    if n%2 == 0:
        n//= 2
    else:
        n = 3*n+1
    dem += 1
    if n > lon_nhat:
        lon_nhat = n
print("So buoc:", dem)
print("Gia tri lon nhat:", lon_nhat)
```

```python
#TH 2.10
import random
import math
random.seed(2026)
for N in (100, 10000, 1000000):
    trong = 0
    for i in range(N):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x * x + y * y <= 1:
            trong += 1
    pi = 4 * trong / N
    sai_so = abs(pi - math.pi)
    print("N =", N)
    print(f"Pi uoc luong = {pi:.6f}")
    print(f"Sai so tuyet doi = {sai_so:.6f}")
    print()
    assert sai_so < 6 / math.sqrt(N)
```

```python
#TH 2.11
import math
# Ham can tim nghiem
def f(x):
    return x**3 - x - 2
# Khoang phan ly nghiem
a = 1
b = 2
sai_so = 1e-10
dem = 0
# Phuong phap chia doi
while b - a >= sai_so:
    m = (a + b) / 2
    if f(a) * f(m) < 0:
        b = m
    else:
        a = m
    dem += 1
# Nghiem gan dung
nghiem = (a + b) / 2
# So buoc ly thuyet
k = math.ceil(math.log2(1 / sai_so))
print(f"Nghiem gan dung: {nghiem:.12f}")
print("So buoc lap:", dem)
print("So buoc ly thuyet:", k)
print(f"Sai so khoang: {b-a:.12e}")
```
