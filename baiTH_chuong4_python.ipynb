```python
# TH 4.1
import math
class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("Ban kinh phai duong")
        self.ban_kinh = ban_kinh
    def dien_tich(self):
        return 4 * math.pi * self.ban_kinh**2
    def the_tich(self):
        return (4 / 3) * math.pi * self.ban_kinh**3
    def __str__(self):
        return f"Hinh cau co ban kinh = {self.ban_kinh}"
hc = HinhCau(2)
print(hc)
print(f"Dien tich: {hc.dien_tich():.3f}")
print(f"The tich: {hc.the_tich():.3f}")
try:
    hc2 = HinhCau(-1)
except ValueError as e:
    print("Loi:", e)
```

```python
# TH 4.3
class SoDuKhongDu(Exception):
    pass
class TaiKhoan:
    def __init__(self, chu_tk, so_du=0):
        if so_du < 0:
            raise ValueError("So du ban dau khong duoc am")
        self.chu_tk = chu_tk
        self.so_du = so_du
        self.lich_su = []
    def nap(self, tien):
        if tien <= 0:
            raise ValueError("So tien nap phai duong")
        self.so_du += tien
        self.lich_su.append(("Nap", tien))
    def rut(self, tien):
        if tien <= 0:
            raise ValueError("So tien rut phai duong")
        if tien > self.so_du:
            raise SoDuKhongDu("So du khong du")
        self.so_du -= tien
        self.lich_su.append(("Rut", tien))
tk = TaiKhoan("An", 1000000)
tk.nap(500000)
tk.rut(200000)
try:
    tk.rut(2000000)
except SoDuKhongDu as e:
    print("Loi:", e)
print("Chu tai khoan:", tk.chu_tk)
print("So du cuoi:", tk.so_du)
print("Lich su giao dich:", tk.lich_su)
assert tk.so_du >= 0
```

```python
# TH 4.5
import math
class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError("Chua cai dat dien tich")
    def __str__(self):
        return "Hinh hoc"
class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r
    def dien_tich(self):
        return math.pi * self.r**2
    def __str__(self):
        return f"Hinh tron (r = {self.r})"
class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def dien_tich(self):
        return self.a * self.b
    def __str__(self):
        return f"Hinh chu nhat ({self.a} x {self.b})"
class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p-self.a) *
                         (p-self.b) * (p-self.c))
    def __str__(self):
        return f"Hinh tam giac ({self.a}, {self.b}, {self.c})"
ds = [HinhTron(1),
      HinhChuNhat(3, 4),
      HinhTamGiac(3, 4, 5)]
ds.sort(key=lambda h: h.dien_tich())
print("Cac hinh theo dien tich tang dan:")
for h in ds:
    print(h, "- Dien tich:", round(h.dien_tich(), 3))
tong = sum(h.dien_tich() for h in ds)
print(f"Tong dien tich: {tong:.6f}")
assert abs(tong - (math.pi + 18)) < 1e-10
```

```python
# TH 4.7
import csv
from collections import defaultdict
gia = defaultdict(list)
with open("goldprice.csv", encoding="utf-8") as f:
    for dong in csv.DictReader(f):
        nam = int(dong["Date"][:4])
        gia[nam].append(float(dong["Price"]))
tb = {}
for nam, ds in gia.items():
    if len(ds) == 12:
        tb[nam] = sum(ds) / len(ds)
tang_giam = {}
for nam in sorted(tb):
    if nam - 1 in tb:
        tang_giam[nam] = (tb[nam] / tb[nam - 1] - 1) * 100
sap_xep = sorted(tang_giam.items(),
                 key=lambda x: x[1], reverse=True)
print("Ba nam tang manh nhat:")
for nam, ti_le in sap_xep[:3]:
    print(f"Nam {nam}: {ti_le:.2f}%")
nam_giam, ti_le_giam = min(tang_giam.items(),
                           key=lambda x: x[1])
print(f"Nam giam manh nhat: {nam_giam}, {ti_le_giam:.2f}%")
assert sap_xep[0][0] == 1980
assert nam_giam == 1981
```

```python
# TH 4.9
import csv
with open("luong.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["nam_kn", "luong"])
    w.writerow([1, 30])
    w.writerow([2, 35])
    w.writerow(["abc", 40])
    w.writerow([3, ""])
    w.writerow([4, 50])
    w.writerow([5, 55])
class DongKhongHopLe(Exception):
    pass
def doc_dong(dong, so_dong):
    try:
        nam_kn = int(dong["nam_kn"])
        luong = float(dong["luong"])
        return (nam_kn, luong)
    except ValueError:
        raise DongKhongHopLe(so_dong)
hop_le = []
dong_loi = []
with open("luong.csv", encoding="utf-8") as f:
    for so_dong, dong in enumerate(csv.DictReader(f), start=2):
        try:
            du_lieu = doc_dong(dong, so_dong)
            hop_le.append(du_lieu)
        except DongKhongHopLe as e:
            dong_loi.append(e.args[0])
luong_tb = sum(luong for nam, luong in hop_le) / len(hop_le)
print("Cac dong hop le:")
for nam, luong in hop_le:
    print("Nam kinh nghiem:", nam, "- Luong:", luong)
print("Danh sach dong loi:", dong_loi)
print("So dong hop le:", len(hop_le))
print("Luong trung binh:", luong_tb)
```
