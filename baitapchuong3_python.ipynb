```python
# TH 3.4
import statistics
ds = [(1, 4.0), (2, 5.5), (3, 6.0),
      (4, 7.5), (5, 8.0)]
x = [a for a, b in ds]
y = [b for a, b in ds]
x_tb = sum(x) / len(x)
y_tb = sum(y) / len(y)
Sxy = sum((xi - x_tb) * (yi - y_tb)
          for xi, yi in ds)
Sxx = sum((xi - x_tb)**2 for xi in x)
b = Sxy / Sxx
a = y_tb - b * x_tb
x_moi = 6
y_du_doan = a + b * x_moi
print("Trung binh x:", x_tb)
print("Trung binh y:", y_tb)
print("Sxy =", Sxy)
print("Sxx =", Sxx)
print("He so a =", a)
print("He so b =", b)
print("Phuong trinh hoi quy: y =", a, "+", b, "* x")
print("Diem du doan khi on 6 gio:", y_du_doan)
ket_qua = statistics.linear_regression(x, y)
print("He so b (statistics):", ket_qua.slope)
print("He so a (statistics):", ket_qua.intercept)
```

```python
# TH 3.6
from collections import defaultdict
ds = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6),
      ("Dung", 5.0), ("Em", 6.0),
      ("Phuc", 3.5), ("Giang", 7.0)]
def diem_chu(diem):
    if diem >= 8.5:
        return "A"
    elif diem >= 7:
        return "B"
    elif diem >= 5.5:
        return "C"
    elif diem >= 4:
        return "D"
    else:
        return "F"
nhom = defaultdict(list)
for ten, diem in ds:
    chu = diem_chu(diem)
    nhom[chu].append(ten)
for chu in sorted(nhom):
    print(chu, ":", nhom[chu])
```

```python
# TH 3.8
ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}
tong = ch1.copy()
for sp, dt in ch2.items():
    tong[sp] = tong.get(sp, 0) + dt
print("Tong doanh thu theo san pham:")
for sp, dt in sorted(tong.items(),
                     key=lambda x: x[1], reverse=True):
    print(sp, ":", dt)
ca_hai = ch1.keys() & ch2.keys()
chi_mot = ch1.keys() ^ ch2.keys()
print("San pham ban o ca hai cua hang:", sorted(ca_hai))
print("San pham chi ban o mot cua hang:", sorted(chi_mot))
```

```python
# TH 3.10
def sang(n):
    con_lai = set(range(2, n + 1))
    for p in range(2, int(n ** 0.5) + 1):
        if p in con_lai:
            con_lai -= set(range(p * p, n + 1, p))
    return con_lai
nt = sang(1000)
print("So luong so nguyen to:", len(nt))
print("Tong cac so nguyen to:", sum(nt))
sinh_doi = []
for p in sorted(nt):
    if p + 2 in nt:
        sinh_doi.append((p, p + 2))
print("So cap nguyen to sinh doi:", len(sinh_doi))
print("Cac cap nguyen to sinh doi:", sinh_doi)
assert len(nt) == 168
assert sum(nt) == 76127
```
